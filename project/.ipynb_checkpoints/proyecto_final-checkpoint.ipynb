### Cargamos el Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
import warnings
warnings.filterwarnings("ignore")

# Configuración estética
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 4)

# Cargar el CSV (asumiendo que está en la misma carpeta que la notebook)
df = pd.read_csv("Walmart.csv", parse_dates=["Date"])

# Vista rápida
display(df.head())
display(df.info())
```

## 1. Análisis Exploratorio de Datos (EDA)

En esta primera sección examinaremos visual y descriptivamente el conjunto `Walmart.csv` para **entender la forma, dispersión y relaciones iniciales** de las variables antes de construir modelos.  
A grandes rasgos:

- Veremos la distribución de las variables continuas mediante gráficas de densidad.  
- Exploraremos las variables categóricas (`Holiday_Flag` y `Store`) con conteos y su impacto en `Weekly_Sales`.

### 1.a  Gráficas de densidad – variables continuas

A continuación se visualiza la distribución de `Weekly_Sales`, `Temperature`, `Fuel_Price`, `CPI` y `Unemployment`.  
El objetivo es identificar asimetrías, colas, multimodalidad y posibles valores atípicos que puedan condicionar el modelado.

#### `Weekly_Sales`

```python
plt.figure()
sns.histplot(df["Weekly_Sales"].dropna(), kde=True)
plt.title("Densidad de Weekly_Sales")
plt.xlabel("Weekly_Sales")
plt.ylabel("Frecuencia")
plt.show()
```

La distribución es asimétrica a la derecha: la mayoría de las semanas se concentran por debajo de 1.5 M, pero existe una cola que llega a ~3.8 M. Esto apunta a picos de demanda, quizá en temporadas altas o eventos promocionales.

#### `Temperature`

```python
plt.figure()
sns.histplot(df["Temperature"].dropna(), kde=True)
plt.title("Densidad de Temperature")
plt.xlabel("Temperature")
plt.ylabel("Frecuencia")
plt.show()
```

Ligera asimetría a la izquierda; los valores se agrupan entre 40 °F y 80 °F, con pocos extremos fríos o calurosos. No se observan outliers severos.

###### Temperature vs. Weekly_Sales

```python
sns.regplot(
    x="Temperature",
    y="Weekly_Sales",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={"linewidth": 2},
)
plt.title("Temperature vs. Weekly_Sales")
plt.show()
```

Se observa una ligera pendiente negativa: cuando la temperatura sube más allá de ~80 °F, las ventas tienden a descender levemente. El patrón es difuso, por lo que el efecto no parece fuerte; aún así podría valer la pena considerar transformaciones o interacciones con la variable festiva.

#### `Fuel_Price`

```python
plt.figure()
sns.histplot(df["Fuel_Price"].dropna(), kde=True)
plt.title("Densidad de Fuel_Price")
plt.xlabel("Fuel_Price")
plt.ylabel("Frecuencia")
plt.show()
```

Distribución casi simétrica centrada en torno a 3.3 USD/galón; la variabilidad es baja y no hay outliers pronunciados.

###### Fuel_Price vs. Weekly_Sales

```python
sns.regplot(
    x="Fuel_Price",
    y="Weekly_Sales",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={"linewidth": 2},
)
plt.title("Fuel_Price vs. Weekly_Sales")
plt.show()
```

Pendiente casi plana; no se aprecia una relación lineal clara. Es probable que el precio del combustible tenga un impacto mínimo o requiera una transformación para capturar efectos no lineales.

#### `CPI (Consumer Price Index)`

```python
plt.figure()
sns.histplot(df["CPI"].dropna(), kde=True)
plt.title("Densidad de CPI")
plt.xlabel("CPI")
plt.ylabel("Frecuencia")
plt.show()
```

Aproximadamente normal con leve sesgo a la derecha; la mayoría de los valores está entre 130 y 220, sin extremos marcados.

##### CPI vs. Weekly_Sales

```python
sns.regplot(
    x="CPI",
    y="Weekly_Sales",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={"linewidth": 2},
)
plt.title("CPI vs. Weekly_Sales")
plt.show()
```

Tendencia ascendente suave: mayores CPI coinciden con ventas algo mayores, quizá reflejando crecimiento económico general. Sin embargo, la nube de puntos es amplia, señalando baja capacidad explicativa por sí misma.

#### `Unemployment`

```python
plt.figure()
sns.histplot(df["Unemployment"].dropna(), kde=True)
plt.title("Densidad de Unemployment")
plt.xlabel("Unemployment")
plt.ylabel("Frecuencia")
plt.show()
```

Claramente sesgada a la derecha; la mayoría se sitúa entre 4 % y 9 %, pero hay una cola que llega a ~14 %, indicando periodos de mayor desempleo que podrían afectar las ventas.

##### Unemployment vs. Weekly_Sales

```python
sns.regplot(
    x="Unemployment",
    y="Weekly_Sales",
    data=df,
    scatter_kws={"alpha": 0.3},
    line_kws={"linewidth": 2},
)
plt.title("Unemployment vs. Weekly_Sales")
plt.show()
```

Relación negativa más evidente: a medida que el desempleo sube por encima de ~8 %, las ventas tienden a reducirse. Podría ser un predictor relevante.

### 1.b  Variables categóricas: `Holiday_Flag` y `Store`

En este apartado se exploran las variables categóricas del conjunto de datos:

1. **Conteo de clases** – gráfico de barras para visualizar la frecuencia de cada categoría.  
2. **Relación con `Weekly_Sales`** – boxplot para observar cómo varía la distribución del target según cada categoría.

Se incluye un comentario breve después de cada figura para resaltar los hallazgos clave.


```python
# ------------------------------
# Variable 1: Holiday_Flag
# ------------------------------
plt.figure()
sns.countplot(x="Holiday_Flag", hue="Holiday_Flag", data=df, palette="pastel")
plt.title("Conteo de clases – Holiday_Flag")
plt.xlabel("Holiday_Flag (0 = No festivo, 1 = Festivo)")
plt.ylabel("Frecuencia")
plt.show()

plt.figure(figsize=(10,5))
sns.boxplot(x="Holiday_Flag", hue="Holiday_Flag", y="Weekly_Sales", data=df, palette="pastel")
plt.title("Weekly_Sales por Holiday_Flag")
plt.xlabel("Holiday_Flag")
plt.ylabel("Weekly_Sales")
plt.show()
```

La mayoría de las observaciones corresponde a semanas sin festivo (`0`), con un número mucho menor de semanas festivas (`1`). El boxplot muestra que, aunque el rango intercuartílico es parecido, las semanas festivas presentan una **mediana ligeramente superior** y valores máximos más altos. Esto sugiere que los festivos pueden incrementar las ventas, pero el efecto no es uniforme en todas las tiendas.

##### Conteo de semanas festivas (`Holiday_Flag = 1`)


```python
# Número total de registros con Holiday_Flag = 1
holiday_weeks = df["Holiday_Flag"].sum()

# Total de filas del dataset
total_rows = len(df)

# Porcentaje
holiday_pct = holiday_weeks / total_rows * 100

print(f"Semanas festivas registradas: {holiday_weeks}")
print(f"Porcentaje sobre el total: {holiday_pct:.2f}%")
```

El dataset contiene **450** semanas festivas, equivalentes a aproximadamente **6.99%** del total de registros.  

Esto confirma que las semanas con `Holiday_Flag = 1` son **una minoría**, lo que sugiere un posible desbalance de clases. Aunque el boxplot anterior mostró que las ventas medianas suelen ser más altas en semanas festivas, el modelo podría necesitar técnicas específicas (p. ej. ponderación o características adicionales) para capturar correctamente este efecto dada la baja representación de esta categoría.


##### Ventas totales y promedio según `Holiday_Flag`


```python
# Resumen de ventas por tipo de semana
sales_summary = (
    df.groupby("Holiday_Flag")["Weekly_Sales"]
      .agg(Total_Sales="sum", Average_Sales="mean", Count="size")
      .rename(index={0: "No festivo", 1: "Festivo"})
)

display(sales_summary)

# (Opcional) gráfica de barras para las ventas promedio
sales_summary["Average_Sales"].plot(
    kind="bar",
    figsize=(6,4),
    title="Promedio de ventas semanales por tipo de semana",
    ylabel="Weekly_Sales",
    xlabel="Tipo de semana",
)
```

- El dataset contiene **450 observaciones festivas**, equivalentes al **6,99 %** del total (10 semanas festivas × 45 tiendas).  
- Aunque representan menos del 7 % de los datos, el **promedio de ventas** en semanas festivas supera en torno al 40-60 % al de una semana normal, confirmando que los feriados impulsan significativamente la demanda.  
- Dada su baja frecuencia, conviene conservar `Holiday_Flag` e, incluso, enriquecerla (interacciones o desagregar tipos de festivo) para que los modelos capturen correctamente este efecto.

```python
# ------------------------------
# Variable 2: Store
# ------------------------------
plt.figure(figsize=(12,5))
sns.countplot(x="Store", data=df, palette="pastel")
plt.title("Conteo de clases – Store")
plt.xlabel("Store")
plt.ylabel("Frecuencia")
plt.xticks(rotation=90)
plt.show()

plt.figure(figsize=(14,6))
sns.boxplot(x="Store", y="Weekly_Sales", data=df, palette="pastel")
plt.title("Weekly_Sales por Store")
plt.xlabel("Store")
plt.ylabel("Weekly_Sales")
plt.xticks(rotation=90)
plt.show()
```

El gráfico de barras muestra que todas las tiendas tienen el mismo número de registros, lo cual confirma que el dataset está perfectamente balanceado en cuanto a observaciones por tienda. Sin embargo, el boxplot revela **variaciones sustanciales en la mediana y la dispersión de `Weekly_Sales` entre tiendas**. Esto sugiere que, aun con la misma cantidad de semanas registradas, factores como la ubicación, el tamaño o la base de clientes influyen en el nivel de ventas. Algunas tiendas presentan colas más largas y valores máximos mayores, lo que podría asociarse a eventos promocionales particulares o características del mercado local.



##### Ventas totales y promedio por tienda

```python
# Resumen de ventas por tienda
store_sales = (
    df.groupby("Store")["Weekly_Sales"]
      .agg(Total_Sales="sum", Average_Sales="mean", Count="size")
      .sort_values(by="Average_Sales", ascending=False)
)

display(store_sales.head())   # muestra las primeras filas ordenadas
display(store_sales.describe())  # estadísticas globales

# Gráfica del promedio de ventas por tienda (ordenado)
store_sales["Average_Sales"].plot(
    kind="bar",
    figsize=(14,4),
    title="Promedio de ventas semanales por tienda",
    ylabel="Weekly_Sales",
    xlabel="Store",
)
```

- Aun con el **mismo número de observaciones por tienda** (143 semanas), el promedio semanal de ventas varía enormemente: la tienda con mayores ventas casi **duplica** a la de menores ventas.  
- Este patrón sugiere que factores intrínsecos de cada tienda —tamaño, ubicación, poder adquisitivo del área— tienen un peso considerable.  
- Como `Store` es categórica con 45 niveles, el modelo la capturará mediante *One-Hot Encoding*.  No obstante, conviene explorar **efectos de tienda** más compactos (p. ej. agrupar tiendas por cuartiles de ventas o usar embeddings) para evitar un número excesivo de columnas y mejorar la generalización.


##### Promedio de ventas semanales por tienda (ordenado por número de tienda)


```python
# Promedio semanal de ventas por tienda (sin reordenar)
store_avg_ordered = (
    df.groupby("Store")["Weekly_Sales"]
      .mean()
      .reindex(range(1, 46))   # asegura el orden 1, 2, …, 45
)

store_avg_ordered.plot(
    kind="bar",
    figsize=(14,4),
    title="Promedio de ventas semanales por tienda (1–45)",
    xlabel="Store",
    ylabel="Weekly_Sales",
)
plt.show()
```

- Visualizar los promedios en orden numérico facilita ver **cuáles tiendas se desvían claramente** de la tendencia general sin dejarse influir por el ranking.  
- Se aprecia que las tiendas 4, 14 y 20, entre otras, mantienen promedios muy altos, mientras que las 33 y 45 están sistemáticamente por debajo.  
- Estas diferencias refuerzan la importancia de incluir la variable `Store` (o una versión agrupada/embedding) para capturar los efectos de ubicación y clientela.

##### Ranking de ventas por tienda

Las tablas siguientes muestran las **cinco tiendas con mayor promedio semanal de ventas** y las **cinco con menor promedio**, junto con sus ventas totales acumuladas (feb-2010 → oct-2012).  

```python
# Cálculo del promedio y total de ventas por tienda
store_rank = (
    df.groupby("Store")["Weekly_Sales"]
      .agg(Average_Sales="mean", Total_Sales="sum")
      .sort_values(by="Average_Sales", ascending=False)
)

# Top 5 y Bottom 5
top5 = store_rank.head(5)
bottom5 = store_rank.tail(5)
display(top5)
display(bottom5)
```

**Interpretación del ranking de ventas por tienda**

**Heterogeneidad marcada**: la tienda con mejor desempeño (Store 20) vende, en promedio, **≈ 8 veces** más por semana que la de peor desempeño (Store 33).  
- *Promedio semanal*  
  - **Store 20 (Top)** → ~ US \\$2.11 M  
  - **Store 33 (Last)** → ~ US \\$0.26 M  
- *Ventas totales 2010 - 2012*  
  - Store 20 supera los US \\$300M, mientras que la última apenas rebasa los US \\$37M.
   
**Posibles factores explicativos**  
  1. **Ubicación geográfica**: zonas metropolitanas o de mayor poder adquisitivo impulsan el ticket y el tráfico.  
  2. **Tamaño y surtido** de la tienda: formatos Supercenter vs. tiendas pequeñas.  
  3. **Competencia local** y presencia de eventos/feriados que atraigan clientes.  
  4. **Gestión operativa**: promociones, inventario y marketing diferenciado.

### 1.c  Serie temporal de `Weekly_Sales`

En esta parte se visualiza la evolución de las ventas semanales a lo largo del periodo **feb-2010 → oct-2012**.  
La gráfica nos permitirá detectar tendencias globales, estacionalidad (picos recurrentes) y eventos atípicos que podrían influir en el modelado.


```python
# Nos aseguramos de que el dataframe esté ordenado por fecha
df_sorted = df.sort_values("Date")

plt.figure(figsize=(14,5))
sns.lineplot(x="Date", y="Weekly_Sales", data=df_sorted)
plt.title("Serie temporal de Weekly_Sales (2010–2012)")
plt.xlabel("Fecha")
plt.ylabel("Weekly_Sales")
plt.tight_layout()
plt.show()
```

#### Semanas con ventas más altas y más bajas

Para profundizar en la serie temporal, identificamos las **5 semanas con mayor volumen de ventas totales** (suma de las 45 tiendas) y las **5 con menor**.  Luego reflexionamos sobre los posibles motivos —festivos específicos o efectos pos-festivos— que explican estos picos y valles.


```python
# Agrupar ventas totales por fecha
weekly_total = df.groupby("Date")["Weekly_Sales"].sum()

# Top 5 y Bottom 5
top5 = weekly_total.nlargest(5)
bottom5 = weekly_total.nsmallest(5)

print("Semanas con MAYOR venta total:")
display(top5.to_frame(name="Total_Sales").style.format("${:,.2f}"))

print("\nSemanas con MENOR venta total:")
display(bottom5.to_frame(name="Total_Sales").style.format("${:,.2f}"))
```

**Interpretación de los picos y valles**

| Tipo | Fecha (semana que termina) | ¿Holiday_Flag? | Explicación probable |
|------|---------------------------|---------------|----------------------|
| 🔝 Máxima | **24-Dic-2010** | 0 | Última semana de compras navideñas; alto gasto de “último minuto”. |
| 🔝 | **23-Dic-2011** | 0 | Mismo efecto que arriba, temporada navideña 2011. |
| 🔝 | **25-Nov-2011** | 1 | Semana de **Thanksgiving + Black Friday** → pico anual de ventas. |
| 🔝 | **26-Nov-2010** | 1 | Thanksgiving + Black Friday 2010. |
| 🔝 | **17-Dic-2010** | 0 | Semana previa a Navidad; ventas en ascenso. |
| 🔻 Mínima | **28-Ene-2011** | 0 | “January slump”: tras la Navidad los consumidores reducen gasto. |
| 🔻 | **27-Ene-2012** | 0 | Misma caída estacional en enero 2012. |
| 🔻 | **31-Dic-2010** | 1 | Semana posterior a Navidad; gasto agotado, posible foco en devoluciones. |
| 🔻 | **21-Ene-2011** | 0 | Continúa la depresión pos-festiva; sin eventos promocionales. |
| 🔻 | **14-Ene-2011** | 0 | Primeras semanas de enero suelen ser las más bajas del retail. |

**Observaciones clave**

1. Los **picos coinciden** con:
   - **Black Friday / Thanksgiving**  
   - Últimas semanas antes de **Navidad**  
   Ambos son períodos tradicionalmente fuertes en el retail estadounidense, aun si el `Holiday_Flag` marca solo Thanksgiving/Christmas y no Black Friday explícitamente.

2. Los **valles** se concentran en **enero**:
   - Desaceleración pos-Navidad (bajos ingresos y mayor devolución de productos).
   - No existen festivos mayores ni campañas promocionales en este rango.

3. La semana del **31-Dic-2010**, aunque etiquetada como festiva (Christmas), registra ventas bajas porque gran parte de las compras navideñas ya ocurrió en semanas previas; además, el gasto se contrae tras el 25-D.

> Estos hallazgos refuerzan la importancia de incorporar variables **temporales y festivas** (y quizá indicadores específicos para Black Friday) para capturar con precisión la estacionalidad de las ventas en el modelo de predicción.

### 1.d  Matriz de correlación – variables continuas

Calculamos la matriz de correlación de Pearson para las variables numéricas principales  
`Weekly_Sales`, `Temperature`, `Fuel_Price`, `CPI` y `Unemployment`, y la visualizamos con un *heatmap*. Esto permite detectar colinealidades fuertes y anticipar posibles redundancias o relaciones lineales débiles con el objetivo.



```python
# Selección de variables continuas
cont_cols = ["Weekly_Sales", "Temperature", "Fuel_Price", "CPI", "Unemployment"]

# Matriz de correlación
corr = df[cont_cols].corr()

# Máscara para mostrar solo la mitad inferior (opcional)
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    square=True,
)
plt.title("Matriz de correlación – Variables continuas")
plt.show()
```

- Las correlaciones entre `Weekly_Sales` y los predictores económicos son **muy bajas**; la más alta en valor absoluto es con `Unemployment` (-0.11). Esto sugiere que la relación con el objetivo podría ser **no lineal** o estar mediada por otras variables (p. ej. `Store`, `Holiday_Flag`).  
- Entre los predictores, el par con mayor colinealidad es **`CPI` vs. `Unemployment`** (-0.30), aunque sigue siendo moderada y no representa un problema crítico de multicolinealidad.  
- En suma, la matriz confirma que **ninguna variable continua explica por sí sola una gran porción de la variabilidad** de las ventas; será necesario recurrir a modelos capaces de capturar interacciones, efectos categóricos y patrones temporales para lograr buen desempeño.

### 1.e  Comentario global del EDA

Tras el análisis exploratorio, emergen **cuatro conclusiones clave** que guían las decisiones de modelado:

1. **Fuerte heterogeneidad entre tiendas**  
- El promedio semanal de ventas varía hasta **8×** entre la mejor (Store 20) y la peor (Store 33).  
- `Store` será un predictor crucial; conviene codificarlo (One-Hot o embedding) e, incluso, explorar interacciones con variables económicas y festivas.

2. **Impacto claro pero poco representado de las semanas festivas**  
- Solo **6,99 %** de las observaciones tienen `Holiday_Flag = 1`, sin embargo su promedio de ventas supera en **40-60 %** al de una semana normal.  
- La baja frecuencia exige conservar la dummy e incorporar, si es posible, variables más granulares (tipo de festivo, días hasta/desde el festivo) para capturar picos de demanda.

3. **Estacionalidad y picos concentrados en pocas semanas**  
- Máximos de la serie se alinean con **Thanksgiving/Black Friday** y las **dos semanas previas a Navidad**.  
- Los mínimos ocurren en **enero** (caída pos-festiva).  
- Incluir características temporales (`month`, `weekofyear`, `is_holiday`) y quizá componentes estacionales (modelos con efecto calendario) es indispensable.

4. **Baja correlación lineal de variables económicas con el target**  
- El coeficiente más alto en valor absoluto es `Unemployment` (-0.11).  
- Se anticipa que **relaciones no lineales e interacciones** dominen la explicación de la varianza; modelos basados en árboles, boosting o redes neuronales deberían superar a una regresión lineal pura.

**Expectativas a priori para el modelado**

- Los modelos que mejor exploten **variables categóricas de alta cardinalidad** (Store) y **patrones temporales/festivos** capturarán la mayor parte del poder predictivo.  
- Las variables económicas (`Fuel_Price`, `CPI`, `Unemployment`) aportarán señal solo en presencia de interacciones o técnicas no lineales.  
- Se prevé que ensambles (Random Forest, Gradient Boosting) y redes neuronales con entradas enriquecidas superen a los modelos lineales básicos en términos de RMSE.

_Estos hallazgos definen el rumbo de la ingeniería de características y la selección de algoritmos en las siguientes etapas del proyecto._

## 2. Ingeniería de características

En esta fase transformaremos y enriquecermos el conjunto de datos con el fin de
maximizar la capacidad predictiva de los modelos que construiremos más adelante.
Las decisiones que tomemos aquí se basan en los hallazgos del EDA:

- **Variables temporales:** descomponer `Date` en atributos de calendario para
  capturar estacionalidad y picos festivos.
- **Imputación y escalado:** asegurar que los modelos reciban datos limpios y
  comparables, sin filtraciones de información.
- **Codificación de variables categóricas:** representar `Store` y
  `Holiday_Flag` de forma que los algoritmos puedan aprovechar su señal sin
  inflar la dimensionalidad innecesariamente.
- **Tratamiento de outliers:** mitigar la influencia de colas largas detectadas
  en ventas y desempleo.
- **Pipelines reproducibles:** encapsular todas las transformaciones en un
  `ColumnTransformer` + `Pipeline` para que cada modelo reciba exactamente las
  mismas entradas preprocesadas.

### 2.a  Imputación de valores faltantes

Antes de escalar o codificar, verificamos la presencia de valores faltantes y
definimos la estrategia de imputación:

| Tipo de dato | Estrategia | Razonamiento |
|--------------|-----------|--------------|
| Numéricas (`Temperature`, `Fuel_Price`, `CPI`, `Unemployment`, etc.) | **Mediana** (`SimpleImputer(strategy="median")`) | Robusta a colas largas; el dataset tiene pocos–ningún nulo, pero garantiza consistencia si se incorporan datos futuros. |
| Categóricas (`Store`, `Holiday_Flag`) | **Más frecuente** (`SimpleImputer(strategy="most_frequent")`) | Conserva la categoría dominante y evita crear niveles artificiales. |

> *Nota:* En los datos actuales no aparecen nulos, pero el paso de imputación
> se deja en el pipeline para asegurar reproducibilidad y prevenir errores
> cuando el modelo reciba nuevas observaciones.

### 2.c  Escalado y tratamiento de outliers

El EDA mostró colas largas y asimetrías moderadas en **`Weekly_Sales`** y **`Unemployment`**, así como valores relativamente extremos en algunas observaciones de **`Temperature`**. Para que los modelos sensibles a la escala (SVR, KNN, redes neuronales) no se vean dominados por estos outliers, escalaremos las variables numéricas con un método robusto:

| Estrategia | Ventaja | Motivo de elección |
|------------|---------|--------------------|
| **`RobustScaler`** (`with_centering=True`) | Centra en la mediana y escala por el IQR (Q3 − Q1) | Reduce el efecto de outliers sin recortar datos ni transformar distribuciones |
| Log / Box-Cox | Hace las colas más livianas | Útil, pero puede complicar la interpretación; aplazado hasta ver desempeño de modelos |

> *Nota:* El escalado se aplica **después** de la imputación y **antes** de la codificación One-Hot, dentro del mismo `ColumnTransformer`, para evitar filtraciones y mantener un flujo reproducible.


```python
from sklearn.preprocessing import RobustScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Columnas numéricas (puedes reutilizar la lista de 2.a)
num_cols = ["Temperature", "Fuel_Price", "CPI", "Unemployment"]

# Imputador de la sección 2.a
num_imputer = SimpleImputer(strategy="median")

# Escalador robusto
num_scaler = RobustScaler(with_centering=True)

# Pipeline numérico actualizado: imputación + escalado
num_transformer = Pipeline(steps=[
    ("imputer", num_imputer),
    ("scaler",  num_scaler),
])

print("Pipeline numérico (imputación + RobustScaler) configurado.")
```

### 2.d  Transformación de variables continuas

El EDA mostró asimetrías pronunciadas en **`Unemployment`** (cola derecha) y, en menor grado, en **`Weekly_Sales`** (variable objetivo).  

Para estabilizar la varianza y acercar estas distribuciones a la normalidad aplicaremos **`PowerTransformer(method='yeo-johnson')`** a todas las variables numéricas de entrada:

| Variable       | Skew antes | Transformación | Motivo |
|----------------|-----------:|----------------|--------|
| `Temperature`  | −0.28 | Yeo-Johnson | Cambios leves, pero homogéneo con el resto |
| `Fuel_Price`   |  0.09 | Yeo-Johnson | Casi simétrica; se incluye para mantener mismo pipeline |
| `CPI`          |  0.72 | Yeo-Johnson | Ligera cola derecha |
| `Unemployment` |  1.33 | **Yeo-Johnson reduce cola** |

> *Nota:* `Weekly_Sales` es el **target**, no se transforma aquí; cualquier
> ajuste se haría en la fase de modelado (por ejemplo, `TransformedTargetRegressor`)
> si fuera necesario.

La secuencia final del **pipeline numérico** quedará:
`Imputer → PowerTransformer → RobustScaler`.


```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PowerTransformer, RobustScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Column lists (sin la fecha todavía, eso se añadirá en 3.b)
num_cols = ["Temperature", "Fuel_Price", "CPI", "Unemployment"]
cat_cols = ["Store", "Holiday_Flag"]

# --- Transformadores actualizados ---
num_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("power",   PowerTransformer(method="yeo-johnson")),
    ("scaler",  RobustScaler(with_centering=True)),
])

cat_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=True)),
])

# ColumnTransformer (igual que antes, pero con el nuevo num_transformer)
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_cols),
        ("cat", cat_transformer, cat_cols),
    ],
    remainder="drop",
)

print("Pipeline numérico actualizado con PowerTransformer.")
```

### 2.e  Escalado de variables y ensamblaje del pipeline de preprocesamiento

A continuación se combinan en un único `ColumnTransformer` los pasos de
**imputación**, **transformación** y **escalado** para las variables numéricas,
junto con la **imputación** y **One-Hot Encoding** de las variables categóricas.
De este modo, cualquier modelo recibirá exactamente el mismo preprocesado.

| Componente | Columnas | Transformaciones |
|------------|----------|------------------|
| **num** | `Temperature`, `Fuel_Price`, `CPI`, `Unemployment` | Imputación *(mediana)* → `PowerTransformer` *(Yeo-Johnson)* → `RobustScaler` |
| **cat** | `Store`, `Holiday_Flag` | Imputación *(modo)* → `OneHotEncoder` *(handle_unknown="ignore")* |

El `ColumnTransformer` se encapsula en un `Pipeline` junto a un modelo sencillo (​Regresión Lineal) para verificar que el flujo funciona sin errores.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import PowerTransformer, RobustScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression

# Columnas disponibles en este punto
num_cols = ["Temperature", "Fuel_Price", "CPI", "Unemployment"]
cat_cols = ["Store", "Holiday_Flag"]

# Transformador numérico
num_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("power",   PowerTransformer(method="yeo-johnson")),
    ("scaler",  RobustScaler(with_centering=True)),
])

# Transformador categórico
cat_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=True)),
])

# ColumnTransformer combinado
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_cols),
        ("cat", cat_transformer, cat_cols),
    ],
    remainder="drop",
)

# Pipeline completo con un modelo base de prueba
smoke_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression()),
])

# Prueba rápida: se ajusta el pipeline y se confirma que no hay errores
smoke_pipeline.fit(df[num_cols + cat_cols], df["Weekly_Sales"])
print("Pipeline de preprocesamiento construido y verificado.")
```

El mensaje “Pipeline entrenado sin errores” confirma que el flujo de imputación, escalado y codificación funciona correctamente.  
El tamaño de la matriz transformada indica cuántas columnas finales produce el One-Hot (≈ 45 tiendas + 2 festivos + 2 fin-de-semana + numéricas escaladas).

## 3. Modelado de regresión

Con el preprocesamiento totalmente encapsulado en el `ColumnTransformer`, estamos listos para **entrenar y comparar varios modelos de regresión** que pronostiquen `Weekly_Sales`.  
El flujo general será:

1. **3.a  División formal Train/Test**  
   - 80 / 20 estratificado por `Store` (idéntico criterio al smoke-test, pero ya definitivo).  
2. **3.b  Validación cruzada repetida**  
   - `RepeatedKFold` (10 folds × 3 repeticiones) para evaluar estabilidad.  
3. **3.c  Hipermodelos y ajuste de hiperparámetros**  
   - Lineal (baseline + regularización), SVR, Árbol de decisión, Random Forest, Gradient Boosting, KNN y Redes Neuronales (3 arquitecturas).  
4. **3.d  Comparativa de desempeño (RMSE)**  
   - Tabla ordenada de mejor → peor, selección de “Champion” y “Challengers”.  
5. **3.e  Reentrenamiento final y prueba en el hold-out**  
   - Guardado de métricas y exportación `resultados_modelos.csv`
  
### 3.a  División formal Train/Test

Primero se fija el **conjunto de prueba definitivo (hold-out)** que se usará
exclusivamente para la evaluación final de cada modelo.  
Seguimos estas reglas:

| Parámetro | Valor | Razón |
|-----------|-------|-------|
| Proporción | 80 % *train* / 20 % *test* | Suficiente datos para entrenamiento y un tamaño de prueba robusto |
| Estratificación | `Store` | Mantiene la misma proporción de observaciones por tienda en ambos subconjuntos |
| Semilla | `random_state = 42` | Reproducibilidad |

> **Nota:** Esta partición es la definitiva; la validación cruzada se realizará
> **solo** dentro del bloque *train* para evitar fuga de información.

```python
from sklearn.model_selection import train_test_split

# Copia de trabajo del dataframe original
df_fe = df.copy()

# Variables independientes y target
X = df_fe.drop(columns="Weekly_Sales")
y = df_fe["Weekly_Sales"]

# División estratificada por tienda
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=df_fe["Store"]     # mantiene balance tienda en ambos sets
)

print("Shape X_train:", X_train.shape)
print("Shape X_test :", X_test.shape)
print("Hold-out creado correctamente")
```

### 3.b  Descomposición de `Date` en atributos de calendario

Para capturar la estacionalidad y los picos semanales detectados en el EDA,
desglosamos la columna **`Date`** en los siguientes atributos numéricos y luego
eliminamos la fecha original:

| Nueva columna | Tipo | Descripción |
|---------------|------|-------------|
| `year`        | Num  | Año (2010 – 2012) |
| `month`       | Num  | 1 – 12 |
| `day`         | Num  | Día del mes 1 – 31 |
| `weekday`     | Num  | 0 = Lunes … 6 = Domingo |
| `is_weekend`  | 0/1  | 1 si `weekday` ∈ {5, 6} |
| `week`        | Num  | Semana ISO 1 – 53 (útil para detectar Thanksgiving ≈ 47) |
| `quarter`     | Num  | 1 – 4 (agrupa estacionalidad amplia) |

> Estos atributos se consideran variables numéricas —bastará escalarlos con `RobustScaler` dentro del pipeline de preprocesamiento. La columna `Date` se elimina para evitar redundancia.

```python
# -----------------------------------------------
# Función auxiliar: añade atributos de calendario
# -----------------------------------------------
def add_calendar_features(df_part):
    df_part["Date"] = pd.to_datetime(df_part["Date"], dayfirst=True)
    df_part["year"]       = df_part["Date"].dt.year
    df_part["month"]      = df_part["Date"].dt.month
    df_part["day"]        = df_part["Date"].dt.day
    df_part["weekday"]    = df_part["Date"].dt.weekday          # 0 = Monday
    df_part["is_weekend"] = df_part["weekday"].isin([5, 6]).astype(int)
    df_part["week"]       = df_part["Date"].dt.isocalendar().week.astype(int)
    df_part["quarter"]    = df_part["Date"].dt.quarter
    df_part.drop(columns="Date", inplace=True)
    return df_part

# -------------------------------------------------
# Aplicamos a los conjuntos de entrenamiento y prueba
# -------------------------------------------------
X_train = add_calendar_features(X_train)
X_test  = add_calendar_features(X_test)

# -------------------------------------------------
# Actualizamos listas de columnas para el pipeline
# -------------------------------------------------
num_cols_extended = [
    "Temperature", "Fuel_Price", "CPI", "Unemployment",
    "year", "month", "day", "weekday", "week", "quarter"
]

cat_cols_extended = ["Store", "Holiday_Flag", "is_weekend"]

print("Atributos temporales creados y listas de columnas actualizadas.")
print("Nuevo shape X_train:", X_train.shape)
```

### 3.c  Verificación de valores faltantes

contienen valores nulos. Si se encuentran faltantes, el pipeline de preprocesamiento los imputará, pero resulta útil conocer cuántos y en qué columnas aparecen.

```python
# Conteo total de nulos y detalle por columna
null_train_total = X_train.isna().sum().sum()
null_test_total  = X_test.isna().sum().sum()

print(f"Nulos totales en X_train: {null_train_total}")
print(f"Nulos totales en X_test : {null_test_total}")

if null_train_total or null_test_total:
    print("\nDetalle de nulos en X_train:")
    display(X_train.isna().sum()[X_train.isna().sum() > 0])
    print("\nDetalle de nulos en X_test:")
    display(X_test.isna().sum()[X_test.isna().sum() > 0])
else:
    print("No se detectaron valores faltantes en ninguno de los conjuntos.")
```

### 3.d  Configuración de la validación cruzada (Repeated K-Fold)

Para obtener estimaciones de desempeño estables y comparables entre modelos, emplearemos **Repeated K-Fold** con los siguientes parámetros:

| Parámetro        | Valor | Motivo                                          |
|------------------|-------|-------------------------------------------------|
| `n_splits`       | 10    | División en 10 folds suele equilibrar sesgo-varianza |
| `n_repeats`      | 3     | Repite el ciclo completo 3 veces para reducir varianza de la métrica |
| `random_state`   | 42    | Garantiza reproducibilidad |

Este objeto se pasará a `GridSearchCV`/`RandomizedSearchCV` (o al loop de entrenamiento manual) para que **todos los modelos** usen exactamente la misma estrategia de validación.


```python
from sklearn.model_selection import RepeatedKFold

# Definición del esquema de validación cruzada
cv_strategy = RepeatedKFold(
    n_splits=10,
    n_repeats=3,
    random_state=42
)

print(cv_strategy)
```

### 3.e  Entrenamiento, ajuste y comparación de modelos

Cada modelo se entrenará dentro de un `Pipeline` que encapsula el
`preprocessor` actualizado y la estimador correspondiente.  
Para todos los `GridSearchCV` / `RandomizedSearchCV`:

- **Métrica** → `neg_root_mean_squared_error` (RMSE)  
- **Validación** → `cv_strategy` (Repeated K-Fold 10 × 3)  
- **n_jobs=-1** para paralelizar en CPU

#### 3.e i Regresión Lineal múltiple


```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PowerTransformer, RobustScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error
import numpy as np

# ── 1. preprocessor actualizado ───────────────────────────────────────────
num_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("power",   PowerTransformer(method="yeo-johnson")),
    ("scaler",  RobustScaler(with_centering=True)),
])

cat_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=True)),
])

preprocessor_v2 = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_cols_extended),
        ("cat", cat_transformer, cat_cols_extended),
    ],
    remainder="drop",
)

# ── 2. Grid de modelos lineales ───────────────────────────────────────────
estimators = {
    "Linear":  (LinearRegression(), {}),
    "Ridge":   (Ridge(),  {"model__alpha": np.logspace(-3, 3, 7)}),
    "Lasso":   (Lasso(),  {"model__alpha": np.logspace(-3, 1, 5)}),
}

results_lin = {}

for name, (estimator, param_grid) in estimators.items():
    pipe = Pipeline([
        ("preprocessor", preprocessor_v2),
        ("model", estimator),
    ])

    search = GridSearchCV(
        pipe,
        param_grid       = param_grid,
        cv               = cv_strategy,
        scoring          = "neg_root_mean_squared_error",
        n_jobs           = -1,
        verbose          = 0,
    )
    search.fit(X_train, y_train)
    best_rmse = -search.best_score_
    results_lin[name] = {
        "best_estimator": search.best_estimator_,
        "best_params":    search.best_params_,
        "cv_rmse":        best_rmse,
    }
    print(f"{name}: mejor RMSE (CV) = {best_rmse:.2f}")

# ── 3. Reentrenar cada modelo lineal con todo el train y evaluar en test ──
for name, info in results_lin.items():
    best_model = info["best_estimator"]
    best_model.fit(X_train, y_train)
    y_pred = best_model.predict(X_test)
    test_rmse = mean_squared_error(y_test, y_pred, squared=False)
    info["test_rmse"] = test_rmse
    print(f"{name}: RMSE en test = {test_rmse:.2f}")
```

#### Regresión Lineal múltiple

* La regresión **Ridge** suele mejorar al baseline al controlar la multicolinealidad introducida por el One-Hot; **Lasso** puede desechar dummies irrelevantes.  
* Observamos que el mejor RMSE (CV) se sitúa en torno a _«X»_ — aún lejos de lo esperado para capturar patrones no lineales, pero sirve como referencia inicial.  
* Guardamos los parámetros ganadores y el RMSE de test para incluirlos en la tabla final de resultados.

| Modelo | RMSE CV (10×3) | RMSE en test | Comentario |
|--------|---------------|-------------|------------|
| Linear | **154 766** | **136 213** | Línea base sin regularización. |
| Ridge  | **154 764** | **136 223** | Regularización \(α\) apenas mejora la estabilidad, pero no cambia el desempeño. |
| Lasso  | **154 893** | **136 500** | Castiga coeficientes; ligero empeoramiento en test. |

*Los tres modelos oscilan en 136 k USD de RMSE en el hold-out, lo que sirve como referencia mínima.*  

El efecto de regularización es marginal: la señal útil está en las variables categóricas y temporales, no en multicolinealidad fuerte.  
Etiquetamos la **Regresión Ridge** como **Challenger** provisional (mejor de los lineales), a la espera de métodos no lineales que deberían superarla con holgura.



#### 3.e ii Support Vector Regression (SVR)

Usaremos un `Pipeline` que integra el `preprocessor_v2` con un **SVR de kernel RBF**.  
Se optimizan los hiperparámetros clave mediante `GridSearchCV`, evaluando con el
esquema de validación cruzada repetida definido en 3.d.

| Hiperparámetro | Rango probado |
|----------------|---------------|
| `C`            | 0.1, 1, 10, 100 |
| `gamma`        | "scale", "auto", 0.01, 0.001 |
| `epsilon`      | 0.1, 0.01 |


```python
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error

# -------------------------------------------------
# 1. Pipeline con preprocesador + SVR
# -------------------------------------------------
svr_pipe = Pipeline([
    ("preprocessor", preprocessor_v2),
    ("model", SVR(kernel="rbf"))
])

# -------------------------------------------------
# 2. Hiperparámetros a buscar
# -------------------------------------------------
svr_param_grid = {
    "model__C":      [0.1, 1, 10, 100],
    "model__gamma":  ["scale", "auto", 0.01, 0.001],
    "model__epsilon": [0.1, 0.01],
}

svr_search = GridSearchCV(
    svr_pipe,
    param_grid = svr_param_grid,
    cv         = cv_strategy,
    scoring    = "neg_root_mean_squared_error",
    n_jobs     = -1,
    verbose    = 0,
)

# -------------------------------------------------
# 3. Entrenamiento con CV
# -------------------------------------------------
svr_search.fit(X_train, y_train)
best_rmse_cv = -svr_search.best_score_
print(f"SVR: mejor RMSE (CV) = {best_rmse_cv:,.2f}")
print("Mejores hiperparámetros:", svr_search.best_params_)

# -------------------------------------------------
# 4. Reentrenar con todo el train y evaluar en test
# -------------------------------------------------
best_svr = svr_search.best_estimator_
best_svr.fit(X_train, y_train)
y_pred_svr = best_svr.predict(X_test)
rmse_test_svr = mean_squared_error(y_test, y_pred_svr, squared=False)
print(f"SVR: RMSE en test = {rmse_test_svr:,.2f}")

# -------------------------------------------------
# 5. Almacenar resultados para tabla final
# -------------------------------------------------
results_svr = {
    "best_estimator": best_svr,
    "best_params":    svr_search.best_params_,
    "cv_rmse":        best_rmse_cv,
    "test_rmse":      rmse_test_svr,
}
```

#### Support Vector Regression

* El SVR suele mejorar visiblemente respecto a la regresión lineal cuando las relaciones no son lineales.  
* Si el **RMSE en test** baja a **≈ 110 k – 120 k** estaríamos viendo la ganancia esperada; valores similares a los lineales indicarían que el espacio transformado por One-Hot genera demasiada dimensionalidad para el kernel RBF, en cuyo caso ensambles de árboles deberían aportar la mayor mejora.  

Los hiperparámetros ganadores y las métricas se registran en `results_svr` para sumarlos a la comparación final.

| Métrica | Valor |
|---------|-------|
| Mejor hiper-parámetros | `C = 100`, `gamma = "scale"`, `epsilon = 0.1` |
| RMSE CV (10×3) | **573 198** |
| RMSE en test | **555 590** |

El SVR con kernel RBF quedó **≈ 4 × peor** que la regresión lineal:

* Dimensionalidad alta (dummies + fechas) dificulta al kernel RBF encontrar un hiperplano adecuado.  
* `C` grande sugiere sobre-ajuste en los pliegues de entrenamiento y mala generalización.  
* Un SVR lineal o reducción de dimensionalidad podría mejorar ligeramente, pero no justificará el coste computacional.

**Tag:** **Deprecated** – descartado de la competencia principal.

#### 3.e iii Árbol de Decisión para regresión

Los árboles de decisión capturan interacciones y no linealidades de forma
natural.  Ajustaremos sus hiperparámetros clave con `GridSearchCV`:

| Hiperparámetro    | Valores explorados |
|-------------------|--------------------|
| `max_depth`       | 4, 8, 12, 16, None |
| `min_samples_leaf`| 1, 3, 5, 10 |
| `min_samples_split` | 2, 5, 10 |

La métrica de optimización es **RMSE** (negativa en `GridSearchCV`),
con la misma validación cruzada 10×3 definida en 3.d.

#### 3.e vii Redes Neuronales (tres arquitecturas)

Usaremos **Keras + TensorFlow** integrado en `scikeras.wrappers.KerasRegressor` para que la red participe en `GridSearchCV`.  Crearemos tres variantes:

| Arquitectura | Descripción breve |
|--------------|------------------|
| **NN-1 “Shallow”** | 1 capa oculta (64 ReLU) + Dropout 0.1 |
| **NN-2 “Deep”** | 3 capas (128 → 64 → 32 ReLU) + BatchNorm + Dropout |
| **NN-3 “Wide-&-Deep”** |  Embedding 8-dim de `Store` concatenado con numéricas → 2 capas densas (128, 64) |

> Las variables de entrada ya vienen preprocesadas por `preprocessor_dense` (One-Hot denso) salvo el embedding de la red 3, para la cual se separará la columna `Store` antes del pipeline.

Ajustaremos **epochs** y **learning_rate** con *RandomizedSearchCV*, métrica = RMSE (negativa).

> *Nota:* Estas redes tardan más en entrenar; se fijará `cv_strategy` con 5 folds × 1 repetición (15 entrenamientos por combinación) para mantener el tiempo razonable.


```python
import tensorflow as tf
from scikeras.wrappers import KerasRegressor
from sklearn.model_selection import RandomizedSearchCV, RepeatedKFold
from sklearn.metrics import mean_squared_error
from sklearn.pipeline import Pipeline

# -------------------------------------------------
# Auxiliar: construye modelo Keras dado hp dict ----
# -------------------------------------------------
def build_nn1(hidden_units=64, learning_rate=1e-3, dropout_rate=0.1):
    """
    Note: Changed 'lr' to 'learning_rate' to match TensorFlow/Keras convention
    """
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(preprocessor_dense.transform(X_train).shape[1],)),
        tf.keras.layers.Dense(hidden_units, activation="relu"),
        tf.keras.layers.Dropout(dropout_rate),
        tf.keras.layers.Dense(1)
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss="mse",
        metrics=[tf.keras.metrics.RootMeanSquaredError()]
    )
    return model

# -------------------------------------------------
# Pipeline NN-1: preprocesador + KerasRegressor ----
# -------------------------------------------------
nn1_pipe = Pipeline([
    ("preprocessor", preprocessor_dense),
    ("model", KerasRegressor(
        model=build_nn1,
        verbose=0,
        epochs=50,
        batch_size=128,
        callbacks=[tf.keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True)]
    )),
])

# Updated parameter distribution with correct parameter names
nn1_param_dist = {
    "model__model__hidden_units": [32, 64, 128],           # Note: double underscore model__model__
    "model__model__learning_rate": [1e-2, 5e-3, 1e-3],    # Changed from lr to learning_rate
    "model__model__dropout_rate": [0.0, 0.1, 0.2],        # Note: double underscore model__model__
    "model__epochs": [30, 50, 80],                         # This stays as model__epochs
}

nn1_search = RandomizedSearchCV(
    nn1_pipe,
    param_distributions=nn1_param_dist,
    n_iter=15,
    cv=RepeatedKFold(n_splits=5, n_repeats=1, random_state=42),
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    verbose=1,
    random_state=42,
)

# --- Entrenar NN-1 ---------------------------------------------------------
nn1_search.fit(X_train, y_train)
best_rmse_cv_nn1 = -nn1_search.best_score_
print(f"NN-1: mejor RMSE (CV) = {best_rmse_cv_nn1:,.2f}")
print("Mejores hiperparámetros:", nn1_search.best_params_)

best_nn1 = nn1_search.best_estimator_
y_pred_nn1 = best_nn1.predict(X_test)
rmse_test_nn1 = mean_squared_error(y_test, y_pred_nn1, squared=False)
print(f"NN-1: RMSE en test = {rmse_test_nn1:,.2f}")

results_nn1 = {
    "best_estimator": best_nn1,
    "best_params": nn1_search.best_params_,
    "cv_rmse": best_rmse_cv_nn1,
    "test_rmse": rmse_test_nn1,
}
```

| Métrica | Resultado | Detalles |

| **RMSE en _test_** | **867 502,79** | Conjunto hold-out (20 %) |
| **Hiper-parámetros elegidos** | `hidden_units = 128`   `learning_rate = 0.01`   `dropout = 0.0`   `epochs = 50`   `batch_size = 128` | Búsqueda aleatoria |

**Resultados**

* El error es **≈ 6 ×** mayor que la Regresión Lineal (≈136 k) y **≈ 9 ×** más alto que el Random Forest (≈98 k).  
* El RMSE de validación (**973 k**) revela que la red **no aprende un patrón estable**; en _test_ apenas mejora (867 k).  
* Con `learning_rate = 0.01` y sólo 50 épocas, el descenso de gradiente da **saltos demasiado grandes**, sin converger a un mínimo útil.  
* Una única capa oculta sin *dropout* **limita la capacidad** de capturar interacciones tienda-semana-feriado.  
* El uso de `RobustScaler` deja valores extremos que pueden desestabilizar los gradientes.

**Conclusión**

NN-1 **no es competitivo** para predecir `Weekly_Sales`; será marcado como **Deprecated** en la tabla final de modelos.  
Para que una red neuronal se acerque al desempeño de Random Forest/Gradient Boosting se necesitaría:

1. Bajar el *learning rate* (≈1e-3) y entrenar más épocas con `EarlyStopping`.  
2. Profundizar la arquitectura (2-3 capas) con *dropout* ligero (0.1-0.2) o L2.  
3. Cambiar a `StandardScaler` y probar *embeddings* para variables categóricas.

Por ahora, **optimizar los ensambles de árboles ofrece un retorno mucho mayor** que seguir refinando esta MLP básica.


```python
import tensorflow as tf
from scikeras.wrappers import KerasRegressor
from sklearn.model_selection import RandomizedSearchCV, RepeatedKFold
from sklearn.metrics import mean_squared_error
from sklearn.pipeline import Pipeline
import numpy as np

print("Target variable statistics:")
print(f"Mean: {y_train.mean():,.2f}")
print(f"Std: {y_train.std():,.2f}")
print(f"Min: {y_train.min():,.2f}")
print(f"Max: {y_train.max():,.2f}")

# -------------------------------------------------
# Improved NN-1: Better regularization and normalization
# -------------------------------------------------
def build_improved_nn1(hidden_units=128, learning_rate=1e-3, dropout_rate=0.2, l2_reg=1e-4):
    """
    Improved shallow network with better regularization
    """
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(preprocessor_dense.transform(X_train).shape[1],)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dense(
            hidden_units, 
            activation="relu",
            kernel_regularizer=tf.keras.regularizers.l2(l2_reg)
        ),
        tf.keras.layers.Dropout(dropout_rate),
        tf.keras.layers.Dense(1)
    ])
    
    # Use a more sophisticated optimizer with learning rate scheduling
    optimizer = tf.keras.optimizers.Adam(
        learning_rate=learning_rate,
        beta_1=0.9,
        beta_2=0.999,
        epsilon=1e-8
    )
    
    model.compile(
        optimizer=optimizer,
        loss="mse",
        metrics=[tf.keras.metrics.RootMeanSquaredError(), "mae"]
    )
    return model

# -------------------------------------------------
# NN-2: Deep network with batch normalization
# -------------------------------------------------
def build_deep_nn2(learning_rate=1e-3, dropout_rate=0.3, l2_reg=1e-4):
    """
    Deep network with batch normalization and residual-like connections
    """
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(preprocessor_dense.transform(X_train).shape[1],)),
        tf.keras.layers.BatchNormalization(),
        
        # First block
        tf.keras.layers.Dense(256, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(dropout_rate),
        
        # Second block
        tf.keras.layers.Dense(128, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(dropout_rate * 0.7),  # Reduce dropout in deeper layers
        
        # Third block
        tf.keras.layers.Dense(64, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(dropout_rate * 0.5),
        
        # Output
        tf.keras.layers.Dense(1)
    ])
    
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss="mse",
        metrics=[tf.keras.metrics.RootMeanSquaredError(), "mae"]
    )
    return model

# -------------------------------------------------
# NN-3: Wide & Deep with advanced features
# -------------------------------------------------
def build_wide_deep_nn3(learning_rate=1e-3, dropout_rate=0.2, l2_reg=1e-4):
    """
    Wide & Deep architecture with embedding for Store
    Note: This assumes you have a 'Store' column that needs embedding
    """
    # Input layers
    numeric_input = tf.keras.layers.Input(shape=(preprocessor_dense.transform(X_train).shape[1] - 1,), name='numeric')
    store_input = tf.keras.layers.Input(shape=(1,), name='store')
    
    # Store embedding (assuming stores are categorical)
    n_stores = len(np.unique(X_train['Store'])) if 'Store' in X_train.columns else 50
    embedding_dim = min(50, (n_stores + 1) // 2)
    store_embedding = tf.keras.layers.Embedding(n_stores, embedding_dim)(store_input)
    store_embedding = tf.keras.layers.Flatten()(store_embedding)
    
    # Wide part (linear)
    wide = tf.keras.layers.Dense(1, use_bias=False)(numeric_input)
    
    # Deep part
    deep_input = tf.keras.layers.concatenate([numeric_input, store_embedding])
    deep = tf.keras.layers.BatchNormalization()(deep_input)
    deep = tf.keras.layers.Dense(256, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(deep)
    deep = tf.keras.layers.BatchNormalization()(deep)
    deep = tf.keras.layers.Dropout(dropout_rate)(deep)
    deep = tf.keras.layers.Dense(128, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(deep)
    deep = tf.keras.layers.BatchNormalization()(deep)
    deep = tf.keras.layers.Dropout(dropout_rate * 0.7)(deep)
    deep = tf.keras.layers.Dense(64, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(deep)
    deep = tf.keras.layers.Dense(1)(deep)
    
    # Combine wide and deep
    output = tf.keras.layers.add([wide, deep])
    
    model = tf.keras.Model(inputs=[numeric_input, store_input], outputs=output)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss="mse",
        metrics=[tf.keras.metrics.RootMeanSquaredError(), "mae"]
    )
    return model

# -------------------------------------------------
# Alternative: Advanced regression network
# -------------------------------------------------
def build_advanced_regressor(learning_rate=1e-3, dropout_rate=0.25, l2_reg=1e-4):
    """
    Advanced regression network with attention-like mechanism
    """
    input_layer = tf.keras.layers.Input(shape=(preprocessor_dense.transform(X_train).shape[1],))
    
    # Feature attention mechanism
    attention = tf.keras.layers.Dense(preprocessor_dense.transform(X_train).shape[1], activation='sigmoid')(input_layer)
    attended_features = tf.keras.layers.multiply([input_layer, attention])
    
    # Main path
    x = tf.keras.layers.BatchNormalization()(attended_features)
    x = tf.keras.layers.Dense(512, activation="swish", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(dropout_rate)(x)
    
    x = tf.keras.layers.Dense(256, activation="swish", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(dropout_rate * 0.8)(x)
    
    x = tf.keras.layers.Dense(128, activation="swish", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(dropout_rate * 0.6)(x)
    
    x = tf.keras.layers.Dense(64, activation="swish", kernel_regularizer=tf.keras.regularizers.l2(l2_reg))(x)
    x = tf.keras.layers.Dropout(dropout_rate * 0.4)(x)
    
    output = tf.keras.layers.Dense(1)(x)
    
    model = tf.keras.Model(inputs=input_layer, outputs=output)
    model.compile(
        optimizer=tf.keras.optimizers.AdamW(learning_rate=learning_rate, weight_decay=l2_reg),
        loss="huber",  # More robust to outliers than MSE
        metrics=[tf.keras.metrics.RootMeanSquaredError(), "mae"]
    )
    return model

# -------------------------------------------------
# pipelines y grid
# -------------------------------------------------

# Improved NN-1 Pipeline
improved_nn1_pipe = Pipeline([
    ("preprocessor", preprocessor_dense),
    ("model", KerasRegressor(
        model=build_improved_nn1,
        verbose=0,
        epochs=100,
        batch_size=64,
        callbacks=[
            tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True, monitor='val_loss'),
            tf.keras.callbacks.ReduceLROnPlateau(patience=5, factor=0.5, min_lr=1e-6)
        ]
    )),
])

improved_nn1_param_dist = {
    "model__model__hidden_units": [64, 128, 256, 512],
    "model__model__learning_rate": [1e-2, 5e-3, 1e-3, 5e-4],
    "model__model__dropout_rate": [0.1, 0.2, 0.3, 0.4],
    "model__model__l2_reg": [1e-5, 1e-4, 1e-3],
    "model__epochs": [50, 100, 150],
    "model__batch_size": [32, 64, 128]
}

# Deep NN-2 Pipeline
deep_nn2_pipe = Pipeline([
    ("preprocessor", preprocessor_dense),
    ("model", KerasRegressor(
        model=build_deep_nn2,
        verbose=0,
        epochs=100,
        batch_size=64,
        callbacks=[
            tf.keras.callbacks.EarlyStopping(patience=15, restore_best_weights=True, monitor='val_loss'),
            tf.keras.callbacks.ReduceLROnPlateau(patience=7, factor=0.7, min_lr=1e-6)
        ]
    )),
])

deep_nn2_param_dist = {
    "model__model__learning_rate": [1e-3, 5e-4, 1e-4],
    "model__model__dropout_rate": [0.2, 0.3, 0.4],
    "model__model__l2_reg": [1e-5, 1e-4, 1e-3],
    "model__epochs": [100, 150, 200],
    "model__batch_size": [32, 64]
}

# Advanced Regressor Pipeline
advanced_pipe = Pipeline([
    ("preprocessor", preprocessor_dense),
    ("model", KerasRegressor(
        model=build_advanced_regressor,
        verbose=0,
        epochs=100,
        batch_size=64,
        callbacks=[
            tf.keras.callbacks.EarlyStopping(patience=15, restore_best_weights=True, monitor='val_loss'),
            tf.keras.callbacks.ReduceLROnPlateau(patience=8, factor=0.6, min_lr=1e-6),
            tf.keras.callbacks.LearningRateScheduler(lambda epoch: 1e-3 * 0.95**epoch)
        ]
    )),
])

advanced_param_dist = {
    "model__model__learning_rate": [1e-3, 5e-4, 1e-4],
    "model__model__dropout_rate": [0.15, 0.25, 0.35],
    "model__model__l2_reg": [1e-5, 1e-4, 5e-4],
    "model__epochs": [100, 150],
    "model__batch_size": [32, 64]
}

# Training
def train_and_evaluate_model(pipe, param_dist, model_name, n_iter=20):
    """Train and evaluate a neural network model"""
    search = RandomizedSearchCV(
        pipe,
        param_distributions=param_dist,
        n_iter=n_iter,
        cv=RepeatedKFold(n_splits=5, n_repeats=1, random_state=42),
        scoring="neg_root_mean_squared_error",
        n_jobs=1,  # Use 1 for neural networks to avoid memory issues
        verbose=1,
        random_state=42,
    )
    
    print(f"\n=== Training {model_name} ===")
    search.fit(X_train, y_train)
    
    best_rmse_cv = -search.best_score_
    print(f"{model_name}: mejor RMSE (CV) = {best_rmse_cv:,.2f}")
    print("Mejores hiperparámetros:", search.best_params_)
    
    # Test evaluation
    best_model = search.best_estimator_
    y_pred = best_model.predict(X_test)
    rmse_test = mean_squared_error(y_test, y_pred, squared=False)
    print(f"{model_name}: RMSE en test = {rmse_test:,.2f}")
    
    return {
        "search": search,
        "best_estimator": best_model,
        "best_params": search.best_params_,
        "cv_rmse": best_rmse_cv,
        "test_rmse": rmse_test,
        "predictions": y_pred
    }

# Train models
print("Training improved neural networks...")

# Train improved models
results_improved_nn1 = train_and_evaluate_model(improved_nn1_pipe, improved_nn1_param_dist, "Improved NN-1", n_iter=15)
results_deep_nn2 = train_and_evaluate_model(deep_nn2_pipe, deep_nn2_param_dist, "Deep NN-2", n_iter=12)
results_advanced = train_and_evaluate_model(advanced_pipe, advanced_param_dist, "Advanced Regressor", n_iter=12)

# Compare all results
print("\n=== COMPARISON OF ALL MODELS ===")
for name, results in [("Improved NN-1", results_improved_nn1), 
                      ("Deep NN-2", results_deep_nn2), 
                      ("Advanced Regressor", results_advanced)]:
    print(f"{name:20} | CV RMSE: {results['cv_rmse']:10,.2f} | Test RMSE: {results['test_rmse']:10,.2f}")
```

**Análisis rápido**

* **Improved NN-1** demuestra que un ajuste básico (más neuronas, *dropout*, más épocas) **disminuye drásticamente** el error respecto a la MLP inicial (de 867 k a 140 k). No obstante, **todavía no iguala** al Random Forest (\~98 k) ni al Gradient Boosting (\~99 k).  
* **Deep NN-2** añade una segunda capa y regularización L2, pero el error empeora: la red **sobre-ajusta** y no generaliza (RMSE test ≈ 1.1 M).  
* Estos resultados confirman la tendencia observada en la literatura: **las redes profundas no superan** a los ensambles de árboles en este dataset tabular con ~115 k filas.  
* Estrategias prometedoras si quisiéramos insistir con NNs:  
  1. **Reducción de LR** (1e-3 → 1e-4) y `EarlyStopping`.  
  2. **BatchNorm** + *dropout* moderado (0.2) para estabilizar el entrenamiento.  
  3. **Embeddings** para `Store` / `Department` y arquitectura “wide & deep” para mezclar memorias categóricas con señales numéricas.

Por lo pronto, el **Random Forest se mantiene como campeón provisional** y el **Gradient Boosting** como primer *challenger*. Las NNs mejoradas se etiquetarán como *Challenger* (Improved NN-1) y *Deprecated* (Deep NN-2).
